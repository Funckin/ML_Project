```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn

print("Checking libraries...\n")

# NumPy
x = np.array([1, 2, 3, 4, 5])
print("NumPy:       OK")
print("  Mean:", np.mean(x))

# Pandas
data = pd.DataFrame({
    "Number": [1, 2, 3, 4, 5],
    "Square": [1, 4, 9, 16, 25]
})
print("Pandas:      OK")
print("  DataFrame created:")
print(data)

# Matplotlib
plt.plot(x, x**2)
plt.title("Matplotlib Test")
plt.xlabel("x")
plt.ylabel("x²")
plt.show()
print("Matplotlib:  OK")

# Scikit-learn
from sklearn.linear_model import LinearRegression

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()
model.fit(X, y)

prediction = model.predict([[6]])

print("Scikit-learn: OK")
print("  Prediction for x=6:", prediction[0])

print("\nAll libraries are working!")
```
